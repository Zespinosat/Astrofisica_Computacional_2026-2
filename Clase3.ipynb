```python
#se importa numpy para el ln
import numpy as np

#damos valor exacto a la integral y las constantes:
Int = 11061.3
u = 2000
m0 = 140000
q = 2100
g = 9.81

#se define la función
def f(t):
    return u * np.log(m0 / (m0 - q *t) ) - g * t 

#se define la regla del trapecio prime
def trapecio(a,b,n):
    h = (b - a) / n
    sum = 0
    for i in range(1,n):
        sum = sum + f(a + i * h)
    return ( h / 2 ) * (f(a) + 2 * sum + f(b))

#se definen los tiempos y se hallan los resultados
t0 = 8
t1 = 30
m = 2
IntTra = trapecio(t0,t1,m)
err_abs = abs(IntTra - Int)
err_rel = 100 * abs(err_abs / Int)

#se imprimen los resultados
print(f"El valor real de la integral es: {Int}")
print(f"La estimación de la integral usando regla del Trapecio es: \n {IntTra:4f}")
print(f"El error absoluto es: \n {err_abs:4f}")
print(f"El error relativo es: \n {err_rel:4f}%")
```

```python
#numpy para la exponencial
import numpy as np

#se define la función y se halla su integral exacta
def f(x):
    return 5 * x * np.exp( - 2 * x)
    
def F(x):
    return  (-5/4) * (2 * x+1)*np.exp(-2*x)


#se define la regla de Simpson 1/3
def simpson1_3(a,b):
    return ((b - a) / 6) * (f(a) + 4 * f( (a + b)/2) + f(b))

#se estima el valor de la integral y sus errores
t0 = 0.1
t1 = 1.3
Int = F(t1) - F(t0)
IntSimp = simpson1_3(t0,t1)
err_abs = abs(IntSimp - Int)
err_rel = 100 * abs(err_abs / Int)

#se imprimen los resultados
print(f"El valor real de la integral es: {Int}")
print(f"La estimación de la integral usando regla del Simpson 1/3 es: \n {IntSimp:4f}")
print(f"El error absoluto es: \n {err_abs:4f}")
print(f"El error relativo es: \n {err_rel:4f}%")
```

```python
#se importa numpy para la exponencial
import numpy as np


#se define la función
def f(x):
  return x - 5 * (1 - np.exp(-x))

print(
      f"{'Iter':<5} | {'xl':<10} | {'xu':<10} | {'xr':<10} | {'ea (%)':<10}"
  )
print("-" * 55)

#método de bisección
def bisection(xl, xu, n):
    #Se calcula internamente la tolerancia
    es = 0.5 * (10 ** (2 - n))
    ea = 100.0 
    xr = xl
    i = 0
    
    while ea > es:
        xr_old = xr
        xr = (xl + xu) / 2.0
        i = i + 1
        #Cálculo del error relativo
        if xr != 0:
            ea = abs((xr - xr_old) / xr) * 100

        #se evalúa el cambio de signo
        if f(xl) * f(xr) < 0:
            xu = xr
        elif f(xl) * f(xr) > 0:
            xl = xr
            
        #Se cálcula el error relativo
        if i > 1:
            ea = abs((xr - xr_old) / xr) * 100
            ea_str = f"{ea:.4f}%"
        else:
            ea_str = "---"

        #Se imprimen los datos de la iteración actual
        print(
            f"{i:<5} | {xl:<10.5f} | {xu:<10.5f} | {xr:<10.5f} | {ea_str:<10}"
        )

    return xr


#Resultados
xl, xu = 0.5, 5
cifras = 3

raiz = bisection(xl, xu, n=cifras)

#se imprimen los resultados
print(
    f"Raíz calculada con {cifras} cifras significativas (Scarborough):"
    f" {raiz:.5f}"
)
```
