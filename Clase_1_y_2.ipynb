```python
#se importa numpy para la exponencial
import numpy as np

#se definen constantes
L_0 = 5 * 10 ** 42
Tau = 111.4

#se define la función sin derivar y su derivada exacta
def L(t):
    return L_0 * np.exp(-t/Tau)

def Lp(t):
    return (-1 * L_0 / Tau) * np.exp(-t/Tau)

#se evalúa en t=0
Lp0 = Lp(0)
print("La derivada exacta en t = 0 es ",Lp0)

#se definen t_0 y h (delta t)
t_0 = 64
h = 2
#se halla la derivada en t_0
Lp64 = Lp(t_0)
print("La derivada exacta en t = 64 es ",Lp64)

#se definen las diferenciales numéricas
def diff_forward(t):
    return (L(t_0 + h) - L(t_0)) / h

def diff_backward(t):
    return (L(t_0) - L(t_0 - h)) / h

def diff_centered(t):
    return (L(t_0 + h) - L(t_0 - h)) / (2 * h)

#se calculan las derivadas
d_forward = diff_forward(t_0)
d_backward = diff_backward(t_0)
d_centered = diff_centered(t_0)

print("La derivada hacía adelante es: ", d_forward)
print("La derivada hacía atrás es: ", d_backward)
print("La derivada centrada es: ", d_centered)



#se calculan los errores relativos
delta_forward = 100 * abs(d_forward-Lp64)/Lp64
delta_backward = 100 * abs(d_backward-Lp64)/Lp64
delta_centered = 100 * abs(d_centered-Lp64)/Lp64

print("El error relativo hacía adelante es: ",delta_forward,"%")
print("El error relativo hacía atrás es: ",delta_backward,"%")
print("El error relativo centrado es: ",delta_centered,"%")
```

```python
#se define la conversión a binarios para enteros
def int_to_bin(n):

    intbin = ""
    
    while n > 0:
        res =  n % 2
        div = n // 2
        intbin = str(res) + intbin
        n =  div
    
    return intbin

#conversión a binarios para fracciones
def frac_to_bin(f, max_bits=20):
    intbin = ""
    fracbin = ""
    num = f
    F_int = int(f)
    f = f - F_int
    
    while F_int > 0:
        res =  F_int % 2
        div = F_int // 2
        intbin = str(res) + intbin
        F_int =  div
    
    while f > 0 and len(fracbin) < max_bits:
        f = f * 2
        bit = int(f)
        fracbin = fracbin + str(bit)
        f = f - bit
    #se imprimen residuos y resultado para verificar 
    #si es exacto
    
    #print(num, "en binario es: \n")
    #print(intbin + "." + fracbin,"\n")
    
    #print("El residuo de", num, "en binario es: \n")
    #print(f,"\n")
    
    return intbin + fracbin

#prueba para verificar funcionalidad de las funciones
print(frac_to_bin(11.1875,4))

#literal a
print(frac_to_bin(18.0308))
print(frac_to_bin(26.0985))
print(frac_to_bin(94.0516))

#literal b
print(frac_to_bin(0.3))

#literal c
print(frac_to_bin(0.1))
print(frac_to_bin(0.2))

#por lo que se puede apreciar que los residuos de los
#números en binario, debido al uso de potencias de dos,
#no necesariamente disminuye ni brinda una representación
#más o menos precisa del númurero décimal, esto debido a
#que es geométricamente imposible representar fracciones
#que no son sumas de potencias de dos exactas.

#literal d, se crea la función

def ieee754(num):
    #Se determina el signo
    if num < 0:
        signo = '1'
        num = -num
    else:
        signo = '0'

    #Se separa parte entera y fraccionaria
    parte_ent = int(num)
    parte_frac = num - parte_ent

    #Se convertierte usando las funciones existentes
    bin_ent = int_to_bin(parte_ent)
    bin_frac = frac_to_bin(parte_frac)

    #Se unen ambas partes
    bina = bin_ent + bin_frac

    #Se normaliza el número
    if parte_ent != 0:
        exponente = len(bin_ent) - 1

        mantisa = bin_ent[1:] + bin_frac

    else:
        posicion = bin_frac.find('1')

        if posicion == -1:
            return '0' * 32

        exponente = -(posicion + 1)
        mantisa = bin_frac[posicion + 1:]

    #Se agregar el bias de IEEE 754
    exponente += 127

    #Se convertierte el exponente a 8 bits
    exponente_bin = format(exponente, '08b')

    #Dado que la mantisa debe tener exactamente 23 bits
    mantisa = (mantisa + '0' * 23)[:23]

    #Se construyen los 32 bits
    print("\n El número ", num, " usando IEEE754 es representado como \n")
    print(signo + exponente_bin + mantisa,"\n")
    
    return signo + exponente_bin + mantisa


ieee754(18.0308)
ieee754(26.0985)
ieee754(94.0516)
ieee754()
```
