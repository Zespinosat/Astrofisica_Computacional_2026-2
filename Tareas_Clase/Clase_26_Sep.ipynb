```python
#importamos math para la exponencial
import math


#Newton-Raphson
def NewtonRaphson(x0, f, f_prima, n=3):
    #Criterio de Scarborough
    es = 0.5 * (10 ** (2 - n))

  
    ea = 100.0  
    x = x0
    i = 0

    #Ya que estamos, se imprimen las iteriaciones (copie y pegue del último ejercicio
    #en clase de la clase anterior
    print(f"{'Iter':<5} | {'E_n (rad)':<12} | {'E_{n+1} (rad)':<12} | {'ea (%)':<10}")
    print("-" * 50)

    while ea > es:
        i += 1
        #Fórmula de Newton-Raphson
        x_next = x - f(x) / f_prima(x)

        #Cálculo del error relativo
        if x_next != 0:
            ea = abs((x_next - x) / x_next) * 100

        print(f"{i:<5} | {x:<12.5f} | {x_next:<12.5f} | {ea:<10.4f}%")

        x = x_next 

    print("-" * 50)
    return x


#Datos del problema
e = 0.7
M = 2.0
E0 = M 


#Función de Kepler
def f(E):
    return E - e * math.sin(E) - M


def f_p(E):
    return 1 - e * math.cos(E)


#Resultados e impresión de estos
raiz = NewtonRaphson(E0, f, f_p, n=3)

print(f"\nLa anomalía excéntrica E es aproximadamente: {raiz:.5f} rad")
```

```python
#numpy para arreglos y logaritmo
import numpy as np

#Datos
t = np.array([0, 10, 20, 30, 40, 50], dtype=float)  # x_i
L = np.array([1.0, 0.914, 0.836, 0.764, 0.699, 0.639], dtype=float)  # y_i

#Transformación logarítmica
z = np.log(L)
n = len(t)

#Cálculo de sumatorias necesarias
sum_x = np.sum(t)
sum_z = np.sum(z)
sum_xz = np.sum(t * z)
sum_x2 = np.sum(t**2)

x_bar = np.mean(t)
z_bar = np.mean(z)

#Cálculo de coeficientes a1 y a0
a1 = (n * sum_xz - sum_x * sum_z) / (n * sum_x2 - (sum_x)**2)
a0 = z_bar - a1 * x_bar

#Resultados e impresión de resultados
A = np.exp(a0)
lamb = -a1

print("a) Constantes de regresión:")
print(f"   A = {A:.6f}")
print(f"   lambda = {lamb:.6f} días^-1")

#Vida media
t_mitad = np.log(2) / lamb
print(f"\nb) Vida media (t_1/2) del 56Co:")
print(f"   t_1/2 = {t_mitad:.2f} días")

#Flujo relativo después de 100 días
L_100 = A * np.exp(-lamb * 100)
print(f"\nc) Flujo relativo a t = 100 días:")
print(f"   L(100) = {L_100:.6f}")
```

```python
#Datos del problema
M_in = 5.0  
tau_dep = 1.5  
R = 0.4  
h = 2.0  # Tamaño de paso (Gy)


# Se define la EDO
def dM_dt(t, M):
    return M_in - ((1 - R) * M) / tau_dep


#Método de Euler
def euler_galaxia(M0=0.0, t0=0.0, h=2.0, tol=5.0):
    t = t0
    M = M0
    i = 0

    print(f"{'Paso':<5} | {'t (Gy)':<8} | {'M(t) (M_sun)':<14} | {'e_rel (%)':<10}")
    print("-" * 45)
    print(f"{i:<5} | {t:<8.1f} | {M:<14.4f} | {'---':<10}")

    while True:
        i += 1
        #Fórmula de Euler
        M_next = M + h * dM_dt(t, M)
        t_next = t + h

        #Error relativo
        e_rel = abs((M_next - M) / M_next) * 100

        print(f"{i:<5} | {t_next:<8.1f} | {M_next:<14.4f} | {e_rel:<10.2f}%")

        t = t_next
        M = M_next

        #e_rel < 5%
        if e_rel < tol:
            break

    print("-" * 45)
    return t, M


#Ejecución
t_final, M_final = euler_galaxia(M0=0.0, t0=0.0, h=h, tol=5.0)
```

```python
#numpy para raíces
import numpy as np

#Datos
Omega0 = 190.0  
K = 1.1e-8 
h = 250.0 
t_target = 2000.0 
n_steps = int(t_target / h)


#EDO
def f(t, Omega):
    return -K * (Omega**3)


#Solución exacta
def Omega_exacta(t):
    return Omega0 / np.sqrt(1 + 2 * K * (Omega0**2) * t)


#Método de Euler
def paso_euler(t, Omega, h):
    return Omega + h * f(t, Omega)


#Método de Runge-Kutta 2 (Punto Medio)
def paso_rk2_punto_medio(t, Omega, h):
    k1 = f(t, Omega)
    k2 = f(t + 0.5 * h, Omega + 0.5 * k1 * h)
    return Omega + k2 * h


#Resultados
t_actual = 0.0
Omega_e = Omega0
Omega_rk = Omega0

print(f"{'Paso':<5} | {'t (yr)':<7} | {'Exacta':<10} | {'Euler':<10} | {'Error E'
    ' (%)':<9} |"f" {'RK2 (P.M.)':<10} | {'Error RK2 (%)':<10}")
print("-" * 80)

#Imprimir condición inicial (Paso 0)
exacta_0 = Omega_exacta(0)
print(f"{0:<5} | {0:<7.0f} | {exacta_0:<10.4f} | {Omega_e:<10.4f} | {0.00:<8.2f}%"
    f" | {Omega_rk:<10.4f} | {0.00:<9.2f}%")

#Iteraciones
for i in range(1, n_steps + 1):
    t_siguiente = i * h
    exacta = Omega_exacta(t_siguiente)

    Omega_e = paso_euler(t_actual, Omega_e, h)
    Omega_rk = paso_rk2_punto_medio(t_actual, Omega_rk, h)
    t_actual = t_siguiente

    #Errores relativos
    err_euler = abs(Omega_e - exacta) / exacta * 100
    err_rk2 = abs(Omega_rk - exacta) / exacta * 100

    print(f"{i:<5} | {t_siguiente:<7.0f} | {exacta:<10.4f} | {Omega_e:<10.4f} |"
      f" {err_euler:<8.2f}% | {Omega_rk:<10.4f} | {err_rk2:<9.2f}%")
```
